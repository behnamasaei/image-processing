```python
from ultralytics import YOLO
```

```python
model = YOLO("yolo26n-seg.pt")
```

```python
import os

DATASET_DIR = os.path.abspath("dataset_pass_segment")

for root, dirs, files in os.walk(DATASET_DIR):
    print(root, "|", len(files), "files")
```

```python

model.train(
    data="dataset_pass_segment/data.yaml",
    epochs=100,
    imgsz=960,
    batch=2,
    device="cpu",
    workers=4,
    patience=20,
    optimizer="AdamW",
    lr0=0.001,
    project="runs/passport",
    name="yolo26n_seg"
)
```
