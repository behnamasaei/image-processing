```python
from ultralytics import YOLO
```

```python
model = YOLO("runs/obb/runs/passport/yolo26_s_v9-5/weights/best.pt")
```

```python
model.train(
    data="data.yaml",          # بعد از اصلاح nc و لیبل‌ها
    epochs=80,
    imgsz=1280,                # برای تست سرعت: 896 یا 800
    batch=4,
    nbs=16,                    # آپدیت‌های بیشتر در هر epoch
    patience=25,
    cos_lr=True,
    device="cpu",              # اگر GPU دارید: device=0
    workers=8,                 # روی CPU نادیده گرفته می‌شود
    cache="ram",

    optimizer="AdamW",
    lr0=0.0005,
    lrf=0.01,
    weight_decay=0.001,
    warmup_epochs=3,

    degrees=10.0, translate=0.08, scale=0.4, shear=1.0, perspective=0.00005,
    fliplr=0.0, flipud=0.0,
    hsv_h=0.015, hsv_s=0.6, hsv_v=0.55,
    mosaic=0.5, close_mosaic=10, mixup=0.0,
    erasing=0.2,

    box=10.0,
    save_period=-1,            # برای صرفه‌جویی در دیسک
    plots=True,
    project="runs/passport",
    name="yolo26_s_v10",
    seed=42,
    deterministic=False,
)
```

```python
import matplotlib.pyplot as plt
```

```python
model = YOLO("runs/obb/runs/passport/yolo26m_obb/weights/best.pt")

results = model.predict(
    source="test3.jpg",
    imgsz=1024,
    conf=0.10,
    iou=0.1,
    save=False,
    show_labels=True,
    show_conf=True,
    line_width=2
)

# Display image with detected OBB boxes
annotated_image = results[0].plot()

plt.figure(figsize=(16, 12))
plt.imshow(annotated_image[:, :, ::-1])
plt.axis("off")
plt.show()
```
