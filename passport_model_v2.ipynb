```python
from ultralytics import YOLO
import torch
import os
```

```python
model = YOLO("yolo26n-obb.pt")
```

```python
os.environ["OMP_NUM_THREADS"] = "20"
os.environ["MKL_NUM_THREADS"] = "20"

torch.set_num_threads(20)
torch.set_num_interop_threads(4)
```

```python
model.train(
    data="data.yaml",

    # Training
    epochs=100,
    imgsz=1280,
    batch=4,
    nbs=16,
    device="cpu",

    # Parallel processing
    workers=8,
    cache="ram",

    # Optimizer
    optimizer="AdamW",
    lr0=0.0005,
    lrf=0.01,
    weight_decay=0.001,
    warmup_epochs=3,
    cos_lr=True,

    # Augmentation
    degrees=10.0,
    translate=0.08,
    scale=0.4,
    shear=1.0,
    perspective=0.00005,

    fliplr=0.0,
    flipud=0.0,

    hsv_h=0.015,
    hsv_s=0.6,
    hsv_v=0.55,

    mosaic=0.5,
    close_mosaic=10,
    mixup=0.0,
    erasing=0.2,

    # Loss
    box=10.0,

    # Performance
    amp=False,
    plots=False,
    save_period=-1,
    deterministic=False,

    # Output
    project="runs/passport",
    name="yolo26_n_v1",
    seed=42,
    resume=True,
)
```

```python
import matplotlib.pyplot as plt
```

```python
model = YOLO("runs/obb/runs/passport/yolo26_n_v1-2/weights/best.pt")

results = model.predict(
    source="dataset_pass_segment/passport_scanned/passport_1.png",
    imgsz=1024,
    conf=0.10,
    iou=0.1,
    save=False,
    show_labels=True,
    show_conf=True,
    line_width=2
)

# Display image with detected OBB boxes
annotated_image = results[0].plot()

plt.figure(figsize=(16, 12))
plt.imshow(annotated_image[:, :, ::-1])
plt.axis("off")
plt.show()
```
