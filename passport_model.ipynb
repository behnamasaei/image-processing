```python
import random, shutil
from pathlib import Path

src_img = Path("raw/images")    # عکس‌های اصلی
src_lbl = Path("raw/labels")    # txt های YOLO
out = Path("dataset")
random.seed(42)

imgs = sorted(p for p in src_img.iterdir() if p.suffix.lower() in {".jpg", ".jpeg", ".png"})
random.shuffle(imgs)
n_val = max(1, int(len(imgs) * 0.2))

for i, img in enumerate(imgs):
    split = "val" if i < n_val else "train"
    (out / "images" / split).mkdir(parents=True, exist_ok=True)
    (out / "labels" / split).mkdir(parents=True, exist_ok=True)
    shutil.copy(img, out / "images" / split / img.name)
    lbl = src_lbl / f"{img.stem}.txt"
    if lbl.exists():
        shutil.copy(lbl, out / "labels" / split / lbl.name)
```

```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")   # مدل pretrained کوچک؛ برای دقت بیشتر yolo11s.pt

model.train(
    data="data.yaml",
    epochs=150,
    imgsz=640,        # برای فیلدهای ریز پاسپورت بهتر از 640 است
    batch=8,
    patience=30,
    fliplr=0.0,       # فلیپ افقی خاموش: متن/چیدمان پاسپورت برعکس نمی‌شود
    degrees=5,        # چرخش کم
    perspective=0.0005,
    hsv_v=0.4,        # تغییر روشنایی
    project="runs",
    name="passport",
)
```

```python
from ultralytics import YOLO

model = YOLO("runs/detect/runs/passport/weights/best.pt")
print(model.val())   # mAP50, mAP50-95

results = model.predict("test.jpg", conf=0.25, save=True)
for box in results[0].boxes:
    print(results[0].names[int(box.cls)], box.xyxy[0].tolist(), float(box.conf))
```

```python
from ultralytics import YOLO
import cv2
from matplotlib import pyplot as plt

model = YOLO("runs/detect/runs/passport/weights/best.pt")

# ارزیابی مدل
metrics = model.val()
print(metrics)

# پیش‌بینی
results = model.predict(
    source="test.jpg",
    conf=0.25,
    save=True,
    show=False
)

# نمایش تصویر با کادرهای تشخیص
annotated_image = results[0].plot()

# تبدیل BGR به RGB برای Matplotlib
annotated_image = cv2.cvtColor(annotated_image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12, 8))
plt.imshow(annotated_image)
plt.axis("off")
plt.show()

```

```python
# نمایش اطلاعات تشخیص‌ها
for box in results[0].boxes:
    class_id = int(box.cls)
    confidence = float(box.conf)

    print(
        f"Class: {results[0].names[class_id]}, "
        f"Coordinates: {box.xyxy[0].tolist()}, "
        f"Confidence: {confidence:.3f}"
    )
```
