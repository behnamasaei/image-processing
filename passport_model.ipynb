```python
import random, shutil
from pathlib import Path

src_img = Path("raw/images")    # عکس‌های اصلی
src_lbl = Path("raw/labels")    # txt های YOLO
out = Path("dataset")
random.seed(42)

imgs = sorted(p for p in src_img.iterdir() if p.suffix.lower() in {".jpg", ".jpeg", ".png"})
random.shuffle(imgs)
n_val = max(1, int(len(imgs) * 0.2))

for i, img in enumerate(imgs):
    split = "val" if i < n_val else "train"
    (out / "images" / split).mkdir(parents=True, exist_ok=True)
    (out / "labels" / split).mkdir(parents=True, exist_ok=True)
    shutil.copy(img, out / "images" / split / img.name)
    lbl = src_lbl / f"{img.stem}.txt"
    if lbl.exists():
        shutil.copy(lbl, out / "labels" / split / lbl.name)
```

```python
import torch
from ultralytics import YOLO
import os

# استفاده از تمام هسته‌های منطقی
os.environ["OMP_NUM_THREADS"] = "20"
os.environ["MKL_NUM_THREADS"] = "20"
os.environ["OPENBLAS_NUM_THREADS"] = "20"

torch.set_num_threads(20)
torch.set_num_interop_threads(4)

print("PyTorch Threads:", torch.get_num_threads())
print("CPU Cores:", os.cpu_count())
```

```python
from ultralytics import YOLO
from pathlib import Path

# مدل قوی‌تر از Nano
model = YOLO("yolo11s.pt")

results = model.train(
    data="raw/data.yaml",

    epochs=30,
    imgsz=640,
    batch=8,

    device="cpu",
    workers=4,
    cache="ram",

    optimizer="AdamW",
    lr0=0.001,
    lrf=0.01,
    weight_decay=0.001,
    warmup_epochs=3,

    fliplr=0.0,
    flipud=0.0,
    degrees=3,
    translate=0.05,
    scale=0.15,
    shear=0.0,
    perspective=0.0003,

    hsv_h=0.015,
    hsv_s=0.3,
    hsv_v=0.2,

    mosaic=0.3,
    mixup=0.0,
    close_mosaic=10,

    patience=20,
    save=True,
    save_period=10,
    val=True,
    plots=True,

    seed=42,
    deterministic=True,

    project="runs",
    name="passport_yolo11s",
)

# مسیر مدل نهایی
best_model_path = (
    Path(model.trainer.save_dir) / "weights" / "best.pt"
)

print("Best model:", best_model_path)
print("Exists:", best_model_path.exists())

# ارزیابی بهترین مدل
best_model = YOLO(str(best_model_path))

metrics = best_model.val(
    data="raw/data.yaml",
    imgsz=640,
    device="cpu",
)

print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
```

```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

model = YOLO("runs/detect/runs/passport_full_cpu/weights/best.pt")

results = model.predict(
    source="test3.jpg",
    conf=0.001,
    iou=0.1,
    save=False,
    verbose=True
)

print("Detected objects:", len(results[0].boxes))

annotated = results[0].plot(
    boxes=True,
    labels=True,
    conf=True,
    line_width=3
)

plt.figure(figsize=(14, 10))
plt.imshow(cv2.cvtColor(annotated, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
for box in results[0].boxes:
    class_id = int(box.cls)
    confidence = float(box.conf)

    print(
        f"Class: {model.names[class_id]}, "
        f"Confidence: {confidence:.4f}, "
        f"Coordinates: {box.xyxy[0].tolist()}"
    )

print("Classes:", model.names)
print("Number of detections:", len(results[0].boxes))
print("Image shape:", results[0].orig_shape)
```
